# NiaPy — visualização genérica de otimização

O notebook é deliberadamente um **thin wrapper**: os detalhes de execução, captura do estado e criação dos gráficos ficam em `visualizer.py`.


### Observação sobre dimensões maiores

Para `DIMENSION > 2`, a animação 3D mostra duas coordenadas selecionadas no plano e o **valor real da função** no eixo vertical. A superfície é uma fatia 2D da função, mantendo as demais coordenadas fixas no melhor ponto final. Para `DIMENSION = 1`, `mode="auto"` usa uma visualização 2D.


```python
# pyright: reportMissingTypeStubs=false

from IPython.display import HTML, display
from niapy.algorithms.basic import CatSwarmOptimization
from niapy.problems import HappyCat

from experiment_specifications import ProblemName
from visualizer import AlgorithmInput, visualize_niapy
```

### Problemas

| Função             | Característica principal          | O que testa                       |
| ------------------ | --------------------------------- | --------------------------------- |
| **Sphere**         | Convexa, unimodal                 | convergência básica               |
| **Rastrigin**      | Multimodal, muitos mínimos locais | exploração vs. exploração local   |
| **Step / Stepint** | Descontínua                       | comportamento em função não suave |


```python
SEED = 27

DIMENSIONS = [10, 100]
MAX_ITERATIONS = 100

PROBLEM_PARAMETERS = {
    "max_iters": MAX_ITERATIONS,
    "mode": "3d",
    "surface_resolution": 80,
    "max_frames": 80,
    "frame_duration_ms": 150,
}
```

```python
def run_optimization(
    algorithm: AlgorithmInput,
    problem: object,
    dimension: int,
):
    result = visualize_niapy(
        **PROBLEM_PARAMETERS,
        algorithm=algorithm,
        problem=problem,
        dimension=dimension,
        reference_points=[[0.0 for _ in range(dimension)]],
    )
    result.optimization_figure.show()
    result.convergence_figure.show()
```

```python
def run_optimizations(
    algorithm: AlgorithmInput,
):
    display(HTML(f"<h3>Dimensões: {DIMENSIONS[0]}<h3/>"))
    display(HTML(f"<h4>Problema: {ProblemName.Rosenbrock}<h4/>"))
    run_optimization(
        algorithm=algorithm, problem=ProblemName.Rosenbrock, dimension=DIMENSIONS[0]
    )
    display(HTML(f"<h4>Problema: {ProblemName.Schwefel}<h4/>"))
    run_optimization(
        algorithm=algorithm, problem=ProblemName.Schwefel, dimension=DIMENSIONS[0]
    )
    display(HTML(f"<h4>Problema: {ProblemName.HappyCat}<h4/>"))
    run_optimization(
        algorithm=algorithm, problem=HappyCat(DIMENSIONS[0]), dimension=DIMENSIONS[0]
    )

    display(HTML(f"<h3>Dimensões: {DIMENSIONS[1]}<h3/>"))
    display(HTML(f"<h4>Problema: {ProblemName.Rosenbrock}<h4/>"))
    run_optimization(
        algorithm=algorithm, problem=ProblemName.Rosenbrock, dimension=DIMENSIONS[1]
    )
    display(HTML(f"<h4>Problema: {ProblemName.Schwefel}<h4/>"))
    run_optimization(
        algorithm=algorithm, problem=ProblemName.Schwefel, dimension=DIMENSIONS[1]
    )
    display(HTML(f"<h4>Problema: {ProblemName.HappyCat}<h4/>"))
    run_optimization(
        algorithm=algorithm, problem=HappyCat(DIMENSIONS[1]), dimension=DIMENSIONS[1]
    )
```

## Cat Swarm


```python
algorithm = CatSwarmOptimization(
    population_size=30,
    mixture_ratio=0.1,
    c1=2.05,
    smp=3,
    spc=True,
    cdc=0.85,
    srd=0.2,
    max_velocity=1.9,
    seed=SEED,
)

run_optimizations(algorithm=algorithm)
```
