# NiaPy — visualização genérica de otimização

O notebook é deliberadamente um **thin wrapper**: os detalhes de execução, captura do estado e criação dos gráficos ficam em `visualizer.py`.


### Observação sobre dimensões maiores

Para `DIMENSION > 2`, a animação 3D mostra duas coordenadas selecionadas no plano e o **valor real da função** no eixo vertical. A superfície é uma fatia 2D da função, mantendo as demais coordenadas fixas no melhor ponto final. Para `DIMENSION = 1`, `mode="auto"` usa uma visualização 2D.


```python
# pyright: reportMissingTypeStubs=false
from niapy.algorithms.basic import CatSwarmOptimization, ParticleSwarmOptimization

from visualizer import visualize_niapy
```

## Particle Swarm


```python
algorithm = ParticleSwarmOptimization(
    population_size=30,
    c1=2.0,
    c2=2.0,
    w=0.7,
    seed=42,
)

result = visualize_niapy(
    algorithm=algorithm,
    problem="Rastrigin",
    dimension=2,
    max_iters=80,
    mode="3d",
    surface_resolution=80,
    max_frames=80,
    frame_duration_ms=150,
    reference_points=[[0.0, 0.0]],
)

result.optimization_figure.show()
result.convergence_figure.show()
```

## Cat Swarm


```python
algorithm = CatSwarmOptimization(
    population_size=30,
    mixture_ratio=0.1,
    c1=2.05,
    smp=3,
    spc=True,
    cdc=0.85,
    srd=0.2,
    max_velocity=1.9,
    seed=42,
)

result = visualize_niapy(
    algorithm=algorithm,
    problem="Rastrigin",
    dimension=2,
    max_iters=80,
    mode="3d",
)

result.optimization_figure.show()
result.convergence_figure.show()
```
